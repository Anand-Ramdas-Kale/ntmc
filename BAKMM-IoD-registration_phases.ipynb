# Registration Phase
    
$\text{Registrationo Authority }(RA)\text{ registers the following entities:}$  
* $\text{drone }(DE_i)$  
* $\text{ground station server }(ES_j)$  
* $\text{cloud server }(CS_k)$
---
---

## Drone Registration

### RSDI1
$$
\begin{align*}
SN_{RA}\;&\colon\;\text{confidential (secret) number of }RA \\
k_{RA}\;&\colon\;\text{confidential key of }RA \\
RID_{RA}\;&\colon\;\text{pseudo identity of }RA
\end{align*}
$$
$RID_{RA}\text{ is calculated as follows: }$
$$
\begin{aligned}
RID_{RA} &= h(ID_{RA}\parallel SN_{RA} \parallel k_{RA}) \\
\end{aligned}
$$

$\text{subsequently it designates - }$


$$
\begin{aligned}
ID_{DE_{i}}\;&\colon\;\text{identifier for } DE_i \\
SN_{DE_{i}}\;&\colon\;\text{secret number for } DE_i \\
k_{DE_{i}}\;&\colon\;\text{private key of } DE_i \\
RID_{DE_{i}}\;&\colon\;\text{pseudo identity for } DE_i \\
RTS_{DE_{i}}\;&\colon\;\text{registration timestamp value of } DE_i \\
TC_{DE_{i}}\;&\colon\;\text{temporal credential parameters for } DE_i \\
TID_{DE_{i}}\;&\colon\;\text{provisional temporary identity for } DE_i \\
\end{aligned}
$$

$\text{the psuedo identity and temporal credentials of }DE_i\text{ are calculated by }RA\text{ as follows: }$

$$
\begin{aligned}
RID_{DE_i}&=h(RID_{RA}\parallel ID_{DE_i}\parallel k_{RA} \parallel k_{DE_i}\parallel SN_{DE_i} \\
TC_{DE_i}&=h(RID_{RA}\parallel ID_{DE_i}\parallel k_{RA}\parallel k_{DE_i}\parallel SN_{DE_i}\parallel RTS_{DE_i})
\end{aligned}
$$

$\text{it generates }TID_{DE_i}\text{. The registration data has subsequently been stored on }DE_i$  

### RSDI2
$DE_i\text{ stores values }\{TID_{DE_i},\;RID_{DE_i},\;TC_{DE_i},\;MS_{DE_i-ES_j},\;h(\cdot)\}$  
$MS_{DE_i-ES_j}\;\colon\;\text{ primary secret key of both }DE_i\text{ and }ES_j$

### Summary table for registration phase of drone

|**$RA$**|**$DE_i$**|
|---|---|
|$\text{Generate }SN_{RA}\;\&\;k_{RA}$||
|$\text{Compute }RID_{RA} = h(ID_{RA}\parallel SN_{RA} \parallel k_{RA})$||
|$\text{Generate }ID_{DE_i}\text{ for }DE_i$||
|$\text{Generate }k_{DE_i}\;\&\;SN_{DE_i}\text{ for }DE_i$||
|$\text{Compute }RID_{DE_i}=h(RID_{RA}\parallel ID_{DE_i}\parallel k_{RA} \parallel k_{DE_i}\parallel SN_{DE_i})$||
|$\text{Compute }TC_{DE_i}=h(RID_{RA}\parallel ID_{DE_i}\parallel k_{RA}\parallel k_{DE_i}\parallel SN_{DE_i}\parallel RTS_{DE_i})$||
|$\text{Generate }TID_{DE_i}$||
||$\text{store }\{TID_{DE_i},\;RID_{DE_i},\;TC_{DE_i},\;MS_{DE_i-ES_j},\;h(\cdot)\}$|

## Ground Station Server Registration

### RSES1
$$
\begin{aligned}
ID_{ES_{j}}\;&\colon\;\text{identifier for } ES_j \\
SN_{ES_{j}}\;&\colon\;\text{secret number for } ES_j \\
k_{ES_{j}}\;&\colon\;\text{private key of } ES_j \\
RID_{ES_{j}}\;&\colon\;\text{pseudo identity for } ES_j \\
RTS_{ES_{j}}\;&\colon\;\text{registration timestamp value of } ES_j \\
TC_{ES_{j}}\;&\colon\;\text{temporal credential parameters for } ES_j \\
TIN_{ES_{j}}\;&\colon\;\text{provisional temporary id number for } ES_j \\
\end{aligned}
$$
$\text{compute pseudo identity and temporal credential parameters of }ES_j$
$$
\begin{aligned}
RID_{ES_j}&=h(RID_{RA}\parallel ID_{ES_j}\parallel k_{RA}\parallel k_{ES_j}\parallel SN_{ES_j}) \\
TC_{ES_j}&=h(RID_{RA}\parallel ID_{ES_j}\parallel k_{RA}\parallel k_{ES_j}\parallel SN_{ES_j}\parallel RTS_{ES_j})
\end{aligned}
$$
$RA\text{ generates }TIN_{ES_j}\text{ a provisional temporary identification number and }MS_{ES_j-CS_k}$
$\text{then }RA\text{ stores the registration info of registered }DE_i\text{ and its own info in its database/memory}$

### RSES2
$\text{Finally }ES_j\text{ contains }$
$$
\{\{(TID_{DE_i},RID_{DE_i})\mid i=1,2,\ldots,n_{DE}\}, TIN_{ES_j},RID_{ES_j},TC_{ES_j},\{MS_{DE_i-ES_j}\mid i=1,2,\ldots,n_{DE}\},MS_{ES_j-CS_k},h(\cdot) \}
$$
$\text{in the region of its secured database, where }n_{DE}\text{ represents the entire quantity of drones deployed under }ES_j$

### Summary table for registration phase of ground server 


|$RA$|$ES_j$|
|---|---|
|$\text{Generate }k_{ES_j}, SN_{ES_j}, ID_{ES_j}\text{ for }ES_j$||
|$\text{Compute }RID_{ES_j}=h(RID_{RA}\parallel ID_{ES_j}\parallel k_{RA}\parallel k_{ES_j}\parallel SN_{ES_j})$||
|$\text{Compute }TC_{ES_j}=h(RID_{RA}\parallel ID_{ES_j}\parallel k_{RA}\parallel k_{ES_j}\parallel SN_{ES_j}\parallel RTS_{ES_j})$||
|$\text{Generate }TIN_{ES_j},\;MS_{ES_j-CS_k}$||
||$\text{Store }\{\{(TID_{DE_i},RID_{DE_i})\mid i=1,2,\ldots,n_{DE}\},$  $TIN_{ES_j},RID_{ES_j},TC_{ES_j},$ $\{MS_{DE_i-ES_j}\mid i=1,2,\ldots,n_{DE}\},MS_{ES_j-CS_k},h(\cdot) \}$|
