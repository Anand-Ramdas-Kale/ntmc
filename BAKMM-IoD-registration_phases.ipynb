# Registration Phase
    
$\text{Registrationo Authority }(RA)\text{ registers the following entities:}$  
* $\text{drone }(DE_i)$  
* $\text{ground station server }(ES_j)$  
* $\text{cloud server }(CS_k)$

## Drone Registration

### RSDI1
$$
\begin{align*}
SN_{RA}\;&\colon\;\text{confidential (secret) number of }RA \\
k_{RA}\;&\colon\;\text{confidential key of }RA \\
RID_{RA}\;&\colon\;\text{pseudo identity of }RA
\end{align*}
$$
$RID_{RA}\text{ is calculated as follows: }$
$$
\begin{aligned}
RID_{RA} &= h(ID_{RA}\parallel SN_{RA} \parallel k_{RA}) \\
\end{aligned}
$$

$\text{subsequently it designates - }$


$$
\begin{aligned}
ID_{DE_{i}}\;&\colon\;\text{identifier for } DE_i \\
SN_{DE_{i}}\;&\colon\;\text{secret number for } DE_i \\
k_{DE_{i}}\;&\colon\;\text{private key of } DE_i \\
RID_{DE_{i}}\;&\colon\;\text{pseudo identity for } DE_i \\
RTS_{DE_{i}}\;&\colon\;\text{registration timestamp value of } DE_i \\
TC_{DE_{i}}\;&\colon\;\text{temporal credential parameters for } DE_i \\
TID_{DE_{i}}\;&\colon\;\text{provisional temporary identity for } DE_i \\
\end{aligned}
$$

$\text{the psuedo identity of }DE_i\text{ is calculated by }RA\text{ as follows: }$

$$
\begin{aligned}
RID_{DE_i}&=h(RID_{RA}\parallel ID_{DE_i}\parallel k_{RA} \parallel k_{DE_i}\parallel SN_{DE_i})
\end{aligned}
$$

$\text{the temporal credential parameters are calculated as:}$  
$$
\begin{aligned}
TC_{DE_i}&=h(RID_{RA}\parallel ID_{DE_i}\parallel k_{RA}\parallel k_{DE_i}\parallel SN_{DE_i}\parallel RTS_{DE_i})
\end{aligned}
$$
$\text{it generates }TID_{DE_i}\text{. The registration data has subsequently been stored on }DE_i$  

### RSDI2
$DE_i\text{ stores values }\{TID_{DE_i},\;RID_{DE_i},\;TC_{DE_i},\;MS_{DE_i-ES_j},\;h(\cdot)\}$  
$MS_{DE_i-ES_j}\;\colon\;\text{ primary secret key of both }DE_i\text{ and }ES_j$
