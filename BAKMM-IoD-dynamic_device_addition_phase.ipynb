# Dynamic Device Addition Phase

## DDA1

$RA\text{ chooses the following things for a new drone }DE_i^{\nu}$

$$
\begin{aligned}
ID_{DE_i}^{\nu}\;&\colon\;\text{id for drone} \\
k_{DE_i}^{\nu}\;&\colon\;\text{secret key for drone} \\
SN_{DE_i}^{\nu}\;&\colon\;\text{secret number for drone}
\end{aligned}
$$

$\text{then it computes the pseudo identity and tempral credentials for drone as}$
$$
\begin{aligned}
RID_{DE_i}^{\nu}&=h(RID_{RA}\parallel ID_{DE_i}^{\nu}\parallel k_{RA}\parallel k_{DE_i}^{\nu}\parallel SN_{DE_i}^{\nu}) \\
TC_{DE_i}^{\nu}&=h(RID_{RA}\parallel ID_{DE_i}^{\nu}\parallel k_{RA}\parallel k_{DE_i}^{\nu}\parallel SN_{DE_i}^{\nu}\parallel RTS_{DE_i}^{\nu})
\end{aligned}
$$
$RA \text{ then chooses:}$
$$
TID_{ED_i}^{\nu}\;\colon\;\text{temporal identity for drone}
$$
$\text{drone stores its registration info into its memory}$
