# Blockchain Implementation Phase

## BIP1

$$
DE_i
\xrightarrow[\;SK_{DE_i,ES_j}\;]{\;Inf_{DE_i}\;}
ES_j
$$
$$
ES_j
\xrightarrow[\;SK_{ES_j,CS_l}\;]{\;PBK_{ES_j}\;}
CS_l
$$

$$
\begin{aligned}
Inf_{DE_i}\;&\colon\;\text{information received from }DE_i \\
PBK_{ES_j}\;&\colon\;\text{partial block created by }ES_j \\
KU_{ES_j}\;&\colon\;\text{public key of }ES_j \\
KS_{ES_j}\;&\colon\;\text{private key of }ES_j \\
tr_x\;&=\;\{tr_1,tr_2,tr_3\ldots,tr_x\} \\
TR_x\;&=E_{KU_{ES_j}}(tr_x) \\
OWI_{ES_j}\;&\colon\;\text{owner's identity }ES_j \\
MT_{root_{ES_j}}\;&\colon\;\text{markle tree root value} \\
PBK_{ES_j}\;&=\;\{OWI_{ES_j},KU_{ES_j},TR_x,MT_{root_{ES_j}}\} \\
\end{aligned}
$$

## BIP2
$CS_l\text{ now generates full block }FBK_{CS_l}$
$$
FBK_{CS_l} = \{BID_{FBK_{CS_l}}, RN_{FBK_{CS_l}}, TSV_{FBK_{CS_l}}, Hash_{FBK_{CS_l}}, Hash_{FBK_{CS_{l-1}}}, OWI_{ES_j},KU_{ES_j},TR_x,MT_{root_{ES_j}}, Sign_{FBK_{CS_l}} \}
$$
$$
\begin{aligned}
BID_{FBK_{CS_l}}\;&\colon\;\text{identification number of the block} \\
RN_{FBK_{CS_l}}\;&\colon\;\text{random nonce value} \\
TSV_{FBK_{CS_l}}\;&\colon\;\text{timestamp} \\
Hash_{FBK_{CS_l}}\;&\colon\;\text{hash of the current block} \\
Hash_{FBK_{CS_{l-1}}}\;&\colon\;\text{hash of the previous block} \\
Sign_{FBK_{CS_l}}\;&\colon\;\text{block's signature} \\
\end{aligned}
$$

## BIP3
