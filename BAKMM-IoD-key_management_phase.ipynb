# Key Management Phase

## AKDEC1
$ES_j\text{ starts communication and produces }$
$$
\begin{aligned}
TS_1\;&\colon\;\text{timestamp value} \\
RS_1\;&\colon\;\text{random secret value}
\end{aligned}
$$
$\text{then it computes:}$
$$
\begin{aligned}
m_1&=h(RS_1\parallel TC_{ES_j}\parallel MS_{ES_j-CS_k}\parallel TS_1)\oplus h(RID_{ES_j}\parallel MS_{ES_j-CS_k}\parallel TS_1) \\
m_2&=h(h(RS_1\parallel TC_{ES_j}\parallel MS_{ES_j-CS_k}\parallel TS_1)\parallel RID_{ES_j}\parallel MS_{ES_j-CS_k}\parallel TS_1)
\end{aligned}
$$
$\text{and sends }$
$$
msg_1=\{TIN_{ES_j},m_1,m_2,TS_1\}
$$
$\text{to }CS_k$

## AKDEC2
$\text{At arrival of }msg_1, CS_k\text{ does the following:}$  
$\text{check }$ $$\lvert TS_1-TS_1^*\rvert\le\Delta T$$  
$\text{if OK, then fetch }$ $$TIN_{ES_j}\rightarrow(RID_{ES_j}, MS_{ES_j-CS_k})$$
$\text{compute: }$
$$
\begin{aligned}
h(RS_1\parallel TC_{ES_j}\parallel MS_{ES_j-CS_k}\parallel TS_1)&=m_1\oplus h(RID_{ES_j}\parallel MS_{ES_j-CS_k}\parallel TS_1) \\
m_2^{'}&=h(h(RS_1\parallel TC_{ES_j}\parallel MS_{ES_j-CS_k}\parallel TS_1)\parallel RID_{ES_j}\parallel MS_{ES_j-CS_k}\parallel TS_1)
\end{aligned}
$$
$\text{check }$ $$m_2^{'}\overset{?}{=}m_2$$
$\text{if OK, then generate: }$
$$
\begin{aligned}
TS_2\;&\colon\;\text{timestamp value} \\
RS_2\;&\colon\;\text{random secret value}
\end{aligned}
$$
$\text{calculate: }$
$$
\begin{aligned}
m_3&=h(RS_2\parallel TC_{CS_k}\parallel MS_{ES_j-CS_k}\parallel TS_2)\oplus h(RID_{ES_j}\parallel MS_{ES_j-CS_k}\parallel TS_1\parallel TS_2) \\
SK_{ES_j,CS_k}&=h(h(RS_2\parallel TC_{CS_k}\parallel MS_{ES_j-CS_k}\parallel TS_2)\parallel h(RS_1\parallel TC_{ES_j}\parallel MS_{ES_j-CS_k}\parallel TS_1)\parallel RID_{ES_j}\parallel MS_{ES_j-CS_k}\parallel TS_1\parallel TS_2) \\
m_4&=h(SK_{ES_j-CS_k}\parallel RID_{ES_j}\parallel MS_{ES_j-CS_k}\parallel T_2) \\
\end{aligned}
$$
$\text{now generate: }$
$$
TIN_{ES_j}^{new}\;\colon\;\text{new temporary identifier for }ES_j
$$

$\text{calculate: }$
$$
\begin{aligned}
m_5&=TIN_{ES_j}^{new}\oplus h(RID_{ES_j}\parallel h(RS_1\parallel TC_{ES_j}\parallel MS_{ES_j-CS_k}\parallel TS_1)\parallel TS_2)
\end{aligned}
$$
$\text{now send}$
$$
\begin{aligned}
msg_2&=\{m_3 , m_4 , m_5 , TS_2\}
\end{aligned}
$$
