# BAKMM-IoD — Resource Guide

## Recommended reading order

1. **SHA-256 / secure hashing**
2. **AES-256**
3. **ECC**
4. **ECDSA**

6. **Blockchain**
7. **pBFT**

## 1. SHA-256 / Cryptographic Hashing

### NIST FIPS 180-4 — Secure Hash Standard
**Primary source.** Covers SHA-2 including SHA-256, message preprocessing, and hash computation.

[ NIST FIPS 180-4 — Secure Hash Standard ](https://csrc.nist.gov/pubs/fips/180-4/upd1/final)

[ NIST FIPS 180-4 — PDF ](https://csrc.nist.gov/files/pubs/fips/180-4/final/docs/fips180-4.pdf)

### NIST — SHA-256 glossary
Short reference for what SHA-256 is and what a hash digest is used for.

[ NIST SHA-256 glossary ](https://csrc.nist.gov/glossary/term/sha_256)

**BAKMM-IoD relevance:** The paper uses `h(·)` extensively for identities, temporary credentials, authentication values, and session-key construction. Its communication-cost analysis uses a 256-bit SHA-256 hash output.

## 2. AES-256

### NIST FIPS 197 — Advanced Encryption Standard
**Primary source.** Official AES specification. Covers AES-128, AES-192, and AES-256.

[ NIST FIPS 197 — AES ](https://csrc.nist.gov/pubs/fips/197/final)

[ NIST AES overview ](https://www.nist.gov/publications/advanced-encryption-standard-aes)

Focus on:

- symmetric-key encryption
- plaintext and ciphertext
- 128-bit block size
- 256-bit AES key
- encryption/decryption
- why AES is used after authentication

**BAKMM-IoD relevance:** The paper explicitly states that AES-256 is used to protect the communication channel after session-key establishment.

## 3. Elliptic Curve Cryptography (ECC)

### NIST SP 800-186 — Elliptic Curve Domain Parameters
**Primary source for ECC domain parameters.**

[ NIST SP 800-186 ](https://csrc.nist.gov/pubs/sp/800/186/final)

Focus on:

- finite fields
- elliptic curves
- points on curves
- point addition
- scalar multiplication
- private/public keys
- elliptic-curve discrete logarithm problem

**BAKMM-IoD relevance:** ECC is used in the blockchain phase to create the ground-station public/private key pair and protect transactions.

## 4. ECDSA

### NIST FIPS 186-5 — Digital Signature Standard
**Primary source for digital signatures and ECDSA.**

[ NIST FIPS 186-5 ](https://csrc.nist.gov/pubs/fips/186-5/final)

Focus on:

- signing key vs verification key
- signature generation
- signature verification
- authenticity/integrity vs confidentiality
- relationship between ECC and ECDSA

**BAKMM-IoD relevance:** The paper states that signatures contained in blocks are verified using ECDSA.

## 5. Merkle Trees

### Ralph Merkle — authentication-tree work
The original Merkle authentication-tree work is a useful historical/primary reference.

[ Ralph Merkle — Method of Providing Digital Signatures ](https://patents.google.com/patent/US4309569A/en)

Focus on:

- leaf hashes
- internal hash nodes
- Merkle root
- Merkle proofs
- why changing one transaction changes the root

**BAKMM-IoD relevance:** The ground station computes `MTreeRoot_ESj` over the transactions and places the root in the partial block.

## 9. Protocol-specific note

The external sources above explain the **primitives**. The BAKMM-IoD paper remains the authority for how those primitives are combined.

For example:

`SHA-256` explains a cryptographic hash, but the paper specifies constructions such as

`RID_DEi = h(...)`

and

`SK_ESj,DEi = h(...)`.

Likewise, AES-256 explains symmetric encryption, while the paper specifies **when the established session key is used to protect communication**.

| XOR | Protocol construction | BAKMM-IoD paper |
| Concatenation `∥` | Protocol construction | BAKMM-IoD paper |

### Not core BAKMM-IoD primitives

HMAC, PUF, fuzzy extractors, and MD5 should **not** be put in the core algorithm list. They appear in the paper's discussion of other schemes/literature rather than as core primitives of the proposed BAKMM-IoD protocol.
