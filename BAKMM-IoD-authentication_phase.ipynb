# Authentication Phase

$\text{Authentication happens between }ES_j\text{ and }DE_i$

## AKDDE1
$DE_i\rightarrow ES_j$
* $DE_i\text{ produces new timestamp value }T_1\text{ and random secret value }rs_1$
* $\text{calculate: }M_1=h(RID_{DE_i}\parallel MS_{DE_i-ES_j}\parallel T_1)\oplus h(TC_{DE_i}\parallel rs_1\parallel MS_{DE_i-ES_j}\parallel T_1)$
* $\text{calculate: }M_2=h(h(TC_{DE_i}\parallel rs_1\parallel MS_{DE_i-ES_j}\parallel T_1)\parallel RID_{DE_i}\parallel T_1)$
* $\text{send msg: }MSG_1=\{TID_{DE_i},M_1,M_2,T_1 \}\text{ through insecure open channel}$

## AKDDE2
$\text{At arrival of }MSG_1\text{, }ES_j\text{ does the following }\colon $
* $\text{check: }\lvert T_1-T_1^*\rvert\le\Delta T$
* $\text{if OK}$
    * $\text{fetch }TID_{DE_i}\rightarrow\{RID_{DE_i}, MS_{DE_i-ES_j}\}$
    * $\text{compute }h(TC_{DE_i}\parallel rs_1\parallel MS_{DE_i-ES_j}\parallel T_1)=M_1\oplus h(RID_{DE_i}\parallel MS_{DE_i-ES_j}\parallel T_1)$
    * $\text{compute }M_2^{'}=h(h(TC_{DE_i}\parallel rs_1\parallel MS_{DE_i-ES_j}\parallel T_1)\parallel RID_{DE_i}\parallel T_1)$
    * $\text{check: }M_2^{'}\overset{?}{=}M_2$
    * $\text{if OK, then }DE_i\text{ is authenticates with }ES_j$
        * $\text{produce a new timestamp }T_2\text{ and random secret value }rs_2\text{ and compute the following}$
        * $M_3=h(RID_{DE_i}\parallel MS_{DE_i-ES_j}\parallel T_2)\oplus h(RID_{ES_j}\parallel TC_{ES_j}\parallel rs_2\parallel MS_{DE_i-ES_j}\parallel T_2)$
        * $SK_{ES_j,DE_i}=h(h(TC_{DE_i}\parallel rs_1\parallel MS_{DE_i-ES_j}\parallel T_1)\parallel h(RID_{ES_j}\parallel TC_{ES_j}\parallel rs_2\parallel MS_{DE_i-ES_j}\parallel T_2)\parallel T_1\parallel T2\parallel RID_{DE_i}\parallel MS_{DE_i-ES_j})$
        * $M_4=h(SK_{ES_j,DE_i}\parallel T_1\parallel T_2\parallel RID_{DE_i})$
        * $M_5=TID_{DE_i}^{new}\oplus h(h(TC_{DE_i}\parallel rs_1\parallel MS_{DE_i-ES_j}\parallel T_1)\parallel RID_{DE_i}\parallel T_2)$
        * $\text{then send }MSG_2=\{M_3,M_4,M_5,T_2\}\text{ to }DE_i \text{through insecure open channel}$

## AKDDE3
$\text{At arrival of }MSG_2,\;DE_i\text{ does the following :}$
* $\text{check: }\lvert T_2-T_2^*\rvert\le\Delta T$
* $\text{if OK}$
    * $\text{compute the following to find session key}$
    * $h(RID_{ES_j}\parallel TC_{ES_j}\parallel rs_2\parallel MS_{DE_i-ES_j}\parallel T_2)=M_3\oplus h(RID_{DE_i}\parallel MS_{DE_i-ES_j}\parallel T_2)$
    * $SK_{ES_j,DE_i}=h(h(TC_{DE_i}\parallel rs_1\parallel MS_{DE_i-ES_j}\parallel T_1)\parallel h(RID_{ES_j}\parallel TC_{ES_j}\parallel rs_2\parallel MS_{DE_i-ES_j}\parallel T_2)\parallel T_1\parallel T2\parallel RID_{DE_i}\parallel MS_{DE_i-ES_j})$
    * $M_4^{'}=h(SK_{ES_j,DE_i}\parallel T_1\parallel T_2\parallel RID_{DE_i})$
    * $\text{check: }M_4^{'}\overset{?}{=}M_4$
    * $\text{if OK, then }ES_j\text{ is authenticated by }DE_i\text{ and computed session key by drone is correct}$
        * $TID_{DE_i}^{new}=M_5\oplus h(h(TC_{DE_i}\parallel rs_1\parallel MS_{DE_i-ES_j}\parallel T_1)\parallel RID_{DE_i}\parallel T_2)$
        * $M_6=h(SK_{ES_j,DE_i}\parallel T_3)$
        * $\text{sends }MSG_3=\{M_6,T_3\}\text{ to }ES_j\text{ through insecure open channel}$

## AKDDE4
$\text{At arrival of }MSG_3, ES_j\text{ does the following :}$
* $\text{check: }\lvert T_3-T_3^*\rvert\le\Delta T$
* $\text{if OK}$
    * $M_6^{'}=h(SK_{ES_j,DE_i}\parallel T_3)$
    * $\text{check: }M_6^{'}\overset{?}{=}M_6$
    * $\text{if OK, then, we assume that session key is correctly established}$

## $\text{Vocabulary}$

$$
\begin{aligned}
T_1&\colon\;\text{new timestamp generated by }DE_i \\
T_2&\colon\;\text{new timestamp generated by }ES_j \\
T_3&\colon\;\text{new timestamp generated by }DE_i \\
T_1^*&\colon\;\text{receiving time of }MSG_1 \\
T_2^*&\colon\;\text{receiving time of }MSG_2 \\
T_3^*&\colon\;\text{receiving time of }MSG_3 \\
\Delta T&\colon\;\text{maximum transmission delay} \\
\end{aligned}
$$

they are using $AES-256$ for symmetric key encryption, for reducing compute cost on drone
